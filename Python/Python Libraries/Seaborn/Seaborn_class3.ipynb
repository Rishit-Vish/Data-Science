# Seaborn – Class 3: Distribution Plots

## What Are Distribution Plots?

Distribution plots show how data values are **spread or concentrated**.

Seaborn offers:
- **`histplot()`** → modern histogram (with optional KDE)
- **`kdeplot()`** → smooth curve of distribution
- **`rugplot()`** → small marks showing individual values  
*(Older `distplot()` is now replaced by `histplot()`.)*


## histplot() – Basic Histogram

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = [10,20,30,40,50,60,70,105,90,100]
sns.histplot(data, kde =True) # kde is that line, kde is optional, you can remove it as well, but for better understanding we use it.
```

## kdeplot() – Kernel Density Estimate

```python
sns.kdeplot(data)
```

## distplot() (Legacy) vs histplot()

> `distplot()` is deprecated.
> Use `histplot()` or `kdeplot()` instead.


## rugplot() – Small Ticks for Observations

```python
# sns.histplot(data,kde=True)
data = [10,20,40,40,50,60,40,70]
sns.rugplot(data, height=0.1)
```

## Multiple Distributions with hue

```python
import pandas as pd
import numpy as np

# Make results reproducible
np.random.seed(42)

# Create 100 boys scores (slightly higher mean)
boys_scores = np.random.normal(loc=75, scale=10, size=100)

# Create 100 girls scores (slightly lower mean)
girls_scores = np.random.normal(loc=70, scale=12, size=100)

# Combine into a dataframe
df = pd.DataFrame({
    "score": np.concatenate([boys_scores, girls_scores]),
    "gender": ["Boy"]*100 + ["Girl"]*100
})

df.head()

# sns.histplot(df, x='score', hue='gender', kde=True) #, So what hue does is - You give it 'gender' so it separates values on the basis of gender,
# # and you will see diff colors on both

sns.kdeplot(df, x='score', hue='gender', fill=True)
```

Key takeaways

histplot() → flexible histogram (can combine with KDE).

kdeplot() → smooth density curve.

rugplot() → quick look at raw values.

distplot() is deprecated → prefer histplot().

Use hue to compare groups in the same distribution plot.
