# Seaborn – Class 2: Categorical Plots


## What Are Categorical Plots?

Categorical plots visualize data where at least one variable is **categorical**  
(e.g., day of week, gender, class).

Seaborn gives several easy options to:
- Compare **counts** or **averages** across categories
- Understand spread & distribution


## barplot() vs countplot()

```python
import seaborn as sns
import pandas as pd

data = {
    "Student": [
        "Ava", "Liam", "Noah", "Emma", "Olivia",
        "Ethan", "Sophia", "Mason", "Isabella", "Lucas"
    ],
    "Gender": [
        "Female", "Male", "Male", "Female", "Female",
        "Male", "Female", "Male", "Female", "Male"
    ],
    "Test_Score": [
        88, 75, 92, 64, 81,
        90, 78, 90, 90, 64
    ]
}

df = pd.DataFrame(data)
# df

# sns.barplot(x='Gender', y='Test_Score', data=df)

# If you use barplot(): It calculates the average score for Boys, It calculates the average score for Girls, Then it draws 2 bars.
# The black line shows: “How sure we are about this average.”, if small line -> stable group, if big line so messy group

# countplot() - just counts how many ?, doesn't calculate avergage and all. It just answers: How many are in each category
sns.countplot(x='Test_Score', data=df)

# That's it, barplot shows average and counplot() - how many?
```

## boxplot() & violinplot()

```python
# Boxplot() shows you how much your data is spread.
# Instead of showwing average (like barplot), it shows Middle value, Spread, High and Low values, weird extreme values.
import seaborn as sns
import pandas as pd

data = {
    "Student": [
        "Ava", "Liam", "Noah", "Emma", "Olivia",
        "Ethan", "Sophia", "Mason", "Isabella", "Lucas"
    ],
    "Gender": [
        "Female", "Male", "Male", "Female", "Female",
        "Male", "Female", "Male", "Female", "Male"
    ],
    "Test_Score": [
        88, 75, 92, 64, 81,
        90, 78, 90, 90, 64
    ]
}
df = pd.DataFrame(data)

# Boxplot()
# sns.boxplot(x='Gender', y='Test_Score', data=df)
# the line inside the box, called 'MEDIAN'.
# Longer Lines - means -> scores are spread more, and shorter lines means -> score are not spreaded more.
# As here Male box is greater/ bigger than Female ones, so we can say that Male test score data is more spreaded.


# violinplot()
sns.violinplot(x='Gender', y='Test_Score', data = df)
# Jaha se mota hai vahi par bahut saare logo ne score kiya hai and top part par kam log hai jinhone ache marks score kiye h, isliye vo patla hai and
# same with bottom part vaha par bhi kam log hai jinhone less marks score kiye hai isliye vo patla h...
# Most values exist in Thicker part of your graph
# if taller graph, so more spreaded data, if wider so Slightly less spreaded data and has more participants at that wider area.
```

## stripplot() & swarmplot()

```python
# stripplot()
sns.stripplot(x='Gender', y='Test_Score', data=df)  # shows every single data point as dot.
# for ex - if some males hsa scored marks like - 75,85,90, so dots will appear on that particular point only not on while plot..

df1 = sns.load_dataset('iris')
df1
sns.swarmplot(x='species', y='petal_length', data=df1)
sns.swarmplot(x='Gender', y='Test_Score', data=df)
# Dots cannot be overlapped

# The only difference between stripplot and swarmplot is - in stripplot dots can overlap and but in swarmplot it doesn't...
```

| Plot        | Best for |
|-------------|-----------|
| **barplot** | Compare average values across categories |
| **countplot** | Show how many items fall in each category |
| **boxplot** | See median, quartiles, and outliers |
| **violinplot** | Same as boxplot + shows distribution |
| **stripplot** | View every observation (good for small data) |
| **swarmplot** | Like stripplot but avoids overlap |
