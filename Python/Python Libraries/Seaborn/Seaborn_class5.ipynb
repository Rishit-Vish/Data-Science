## Why Matrix / Grid Plots?

Matrix or grid-style plots are perfect for:
- Showing correlations between variables  
- Visualizing pivot tables  
- Discovering patterns in large datasets


### Heatmaps (heatmap())

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

data = np.array([[1, 5, 3],
                 [7, 2, 6],
                 [4, 8, 9]])

sns.heatmap(data)
plt.show()
# heatmap represents data into colors to understand it quickly
# heatmap shows data like - Bigger numbers - darker/ warmer colors, Smaller numbers - lighter /cooler colors

df = sns.load_dataset('tips')
df
sns.heatmap(df, numeric_only=True)
plt.show()
```

### Cluster Map (clustermap())

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np # we imported this because we are doing operations on numbers, not on categorical columns

data = np.array([[1,5,3],
                 [7,2,6],
                 [4,8,9]])
sns.clustermap(data)
plt.show()
# it doesn't just color the boxes, but also shows the similar group columns and similar group rows by drawing small tree diagrams (called dendograms)
```

### Styling & Themes

```python
# Heatmap Styling/ Theme
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

df = np.array([[1,5,3],
               [2,8,4],
               [3,5,7]])

# sns.set_theme(style='dark')
# sns.set_palette('viridis')
# annot=True -> means it will show Numbers on graph
# cmap -> color style
# linewidths=1 -> space between cells
# linecolor='white'  -> grid color
# square=True -> sqaure cells
# char=True  -> show color bar

sns.heatmap(df, annot=True, fmt='.2f', cmap='coolwarm',linewidths=1, linecolor='white', square=True, cbar=True)
plt.title('Heatmap made by Rishit', fontsize=14)
plt.xticks(rotation=45)  # this xticks means -> That numbers on Axis
plt.yticks(rotation=80)
plt.show()
```

```python
# Clustemap Styling/ Theme
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt

df = np.array([[1,5,4],
               [9,4,6],
               [7,33,6]])
# sns.clustermap(df, cmap='coolwarm', annot=True, figsize=(8,8), linewidths=1, center=0)
# Adjust dendrogram size
sns.clustermap(df,
               cmap='viridis',
               dendrogram_ratio=(0.2, 0.5),
               cbar_pos = (0.2, 0.8, 0.05, 0.18)
)
```
