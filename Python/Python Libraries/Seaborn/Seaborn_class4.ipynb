# Seaborn – Class 4: Regression & Mixed Plots

## Regression & Mixed Plots

Seaborn provides plots that combine **relationships** and **statistical models** or mix multiple plot types.

- `regplot()` / `lmplot()` → scatter + regression line  
- `jointplot()` → scatter with hist/KDE  
- `pairplot()` → multiple relationships in one go


# regplot() – Scatter + Regression Line

```python
import seaborn as sns
import pandas as pd

data = {
    "hours_studied": [1, 2, 3, 4, 4, 6, 10, 8],
    "exam_score": [50, 55, 65, 70, 75, 85, 90, 95]
}

df = pd.DataFrame(data)

sns.regplot(x='hours_studied', y='exam_score', data=df)
# regplot not only shows the graph, but it draws a line that shows the overall trend,. That line is called regression line.
# That shaded blue zone - We think the real trend is somewhere inside this shaded zone.
# Narrow shade = more sure, Wide shade = less sure
# Dots = real students, Line = overall trend, Shade = how confident the model is
```

# lmplot() – Regression with Facets

```python
data = {
    "hours_studied": [1,2,3,4,5,6,7,8],
    "exam_score": [50,55,65,70,75,85,90,95],
    "gender": ["Boy","Boy","Boy","Boy","Girl","Girl","Girl","Girl"]
}

# Facets -> Split the graph into separate mini-graphs based on a category.

df = pd.DataFrame(data)
# df
sns.lmplot(x='hours_studied', y='exam_score', hue='gender',col='gender', data=df)
```
