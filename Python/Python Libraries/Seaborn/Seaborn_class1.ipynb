## What is Seaborn?

Seaborn is a **high-level data visualization library** built on top of Matplotlib.  
It provides:
- Beautiful, modern default styles  
- Concise syntax (less code for complex visuals)  
- Integration with Pandas DataFrames  

Perfect for quick, attractive plots while keeping full Matplotlib power underneath.


###  What Are Relational Plots?

Relational plots are used to **show the relationship between two variables**.

- They focus on how one variable changes with respect to another.
- Typically used for continuous (numeric) data, but can also mix with categories.

In Seaborn, the main relational plots are:

1. **`scatterplot()`** → displays individual data points  
2. **`lineplot()`** → shows a continuous trend or summary line

> Both functions belong to Seaborn’s *relational* family.


## You can load sample DataSets using SNS

```python
import seaborn as sns

# scatterplot()  - Tells the relation between Two things
# if the dots are upward so the realtion is postive
# If downwards, so relation is negative
# if random -> no clear relation
# -> scatterplot

df = sns.load_dataset("iris")
sns.scatterplot(x = 'sepal_length',y= 'petal_width', data=df)

df.head()
df.tail()
df.shape
df.info()

# lineplot, use 'iris' dataset to see graph in a good way, cleaner way
tips = sns.load_dataset('tips')
sns.lineplot(x='total_bill', y='day', data=tips)
print(tips)
sns.lineplot()
```

### Styling & Color Palettes

```python
df = sns.load_dataset('iris')
sns.scatterplot(x='petal_length', y='sepal_length', data=df,color='red')

df = sns.load_dataset('titanic')
sns.scatterplot(x='class', y='age', data=df, color='green')

df = sns.load_dataset('titanic')
sns.scatterplot(x='who', y='age', hue ='embark_town', data=df)  # if you have categoried then you can use 'hue'
sns.set_style('ticks')  # Can be more like - "white","dark","whitegrid","darkgrid","ticks"
```

```python
df = sns.load_dataset('flights')
sns.scatterplot(x='month', y='passengers', style='year', data=df)

df = sns.load_dataset("flights")
sns.lineplot(x='year', y='passengers', hue='month',data=df)

df = sns.load_dataset('tips')
sns.lineplot(x='total_bill', y='tip', hue='sex',style='day', data=df)
```
