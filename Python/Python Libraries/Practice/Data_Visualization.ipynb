```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data
x = np.arange(0, 10, 1)
y1 = x ** 2
y2 = np.sqrt(x)

# Create subplots with a shared X-axis
fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True, figsize=(8, 6))

# First dataset
ax1.plot(x, y1, color='blue', label='y = x²')
ax1.set_ylabel("Dataset 1")
ax1.set_title("Comparison of Two Datasets")
ax1.legend()
ax1.grid(True)

# Second dataset
ax2.plot(x, y2, color='red', label='y = √x')
ax2.set_xlabel("X-axis")
ax2.set_ylabel("Dataset 2")
ax2.legend()
ax2.grid(True)

plt.tight_layout()
plt.show()
```
