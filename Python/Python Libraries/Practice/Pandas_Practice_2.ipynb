# **Feature Engineering**

```python
import pandas as pd
import random

titles = [
    "Fullmetal Alchemist: Brotherhood","Steins;Gate","Bleach: Thousand Year War",
    "Gintama°","Attack on Titan Season 3 Part 2","Hunter x Hunter (2011)",
    "Code Geass R2","Death Note","Vinland Saga","Demon Slayer",
    "Jujutsu Kaisen","One Piece","Naruto Shippuden","Chainsaw Man",
    "Mob Psycho 100","Spy x Family","Haikyuu!!","Monster",
    "Cowboy Bebop","Black Clover","Blue Lock","Parasyte","Erased",
    "Your Lie in April","Dr. Stone","Psycho-Pass",
    "Neon Genesis Evangelion","Baccano!","Made in Abyss","Kaguya-sama",
    "Re:Zero","Mushoku Tensei","Tokyo Revengers","86 Eighty-Six",
    "JoJo Part 5","Fruits Basket","Odd Taxi","Dororo",
    "Great Teacher Onizuka","Samurai Champloo","Fire Force",
    "No Game No Life","Overlord","The Promised Neverland",
    "Violet Evergarden","Akame ga Kill!","Charlotte",
    "Noragami","Hellsing Ultimate","Blue Exorcist"
]

types = ["TV","Movie","OVA","ONA","Special"]
months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]

rows = []

for rank, title in enumerate(titles, start=1):

    anime_type = random.choice(types)

    eps = 1 if anime_type == "Movie" else random.randint(10,150)

    start_year = random.randint(2005,2023)
    start_month = random.choice(months)

    if anime_type == "Movie":
        end_date = "-"
    else:
        end_year = start_year + random.randint(0,2)
        end_month = random.choice(months)
        end_date = f"{end_month} {end_year}"

    members = f"{random.randint(150000,4500000):,}"

    score = round(random.uniform(7.50,9.30),2)

    title_col = (
        f"{title}"
        f"{anime_type} "
        f"({eps} eps)"
        f"{start_month} {start_year} - {end_date}"
        f"{members} members"
    )

    rows.append([rank, title_col, score])

df = pd.DataFrame(rows, columns=["Rank","Title","Score"])

# print(df.head())
df.head()
```
