```python
# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],  # Student 1
    [19, 92, 88],  # Student 2
    [17, 76, 95],  # Student 3
    [18, 65, 70],  # Student 4
    [20, 90, 85]   # Student 5
])
```

```python
# Give shape
data.shape
```

```python
# Take out the average
np.mean([data[:,0]])
```

```python
# Extract math mark of all student
data[:,1]
```

```python
# science highest mark ?
np.max(data[:,2])
```

```python
# Get details of the student who scored more than 90 in Math
data[data[:,1] > 90]
```

```python
# Increase math mark of all students by 5
data[:,1] +5
```

```python
# find how many students are younger than 19
len(data[data[:,0] < 19])   # we used len() because we need to count
```

```python
# calculate the avg marks in each subject, column-wise mean.
np.mean(data[:,1:], axis=0)
```

```python
# get data of students who scored atleast 80 in both subjects
data[ (data[:,1] >= 80) & (data[:,2] >= 80)]
```
