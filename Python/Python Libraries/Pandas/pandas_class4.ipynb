## 1. Boolean Indexing & Conditional Selection
- Use comparison operators to build conditions (e.g., `df['col'] > 50`).
- Pass the condition inside `df[...]` to get matching rows.
- Combine multiple conditions with `&` (AND), `|` (OR), and `~` (NOT).
- Parentheses are required when combining conditions.

```python
import pandas as pd
import numpy as np
a = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Age": [25, 30, 35, 40, 22],
    "Score": [88, 92, 79, 85, 95]
}

b = pd.DataFrame(a)

print(b,"\n\n")

# # Select rows where age is greater than 30
# c = b[b['Age']>30]

# # rows select where Score is less than 90
# d = b[b['Score'] < 90]


# # Using AND(&)
# e = b[(b['Age'] > 30) & (b['Score'] < 90)]


# # Using same cond change the operator (|)
# f = b[(b['Age'] > 30) | (b['Score'] < 90)]

# # we want score which is not greater than 85
# g = b[~(b['Score'] > 85)]  # Using NOT ~ operator
```

## 2. Filtering Rows with Multiple Conditions
- Combine two or more conditions to refine selection.
- Remember to enclose each condition in parentheses.

```python
# Inclusive  (Cond must be in parentheses)
ab = b[(b['Age'] >= 20) & (b['Age'] <= 30)]
ab
```

## 3. Using `query()` for Cleaner Filtering
- `query()` allows filtering rows with a string expression.

```python
# Age should be greater than 20 using query()
# b.query("Age<20")

# Multiple Conditions - Age < 35 and Score >= 85
# b.query("Age < 35 and Score >= 85")

# Age >= 20 and Age <=30
# b.query("Age >=20 & Age <= 30")
```
