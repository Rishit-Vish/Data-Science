## 1. Converting Columns to Datetime (`pd.to_datetime`)
- Use `pd.to_datetime()` to convert strings or numbers into datetime objects.
- Once converted, Pandas recognizes them as dates, enabling powerful operations.

```python
import pandas as pd

# 1. Create the dataset
data = {
    "OrderID": [1001, 1002, 1003, 1004, 1005],
    "OrderDate": ["2024-01-05", "2024/01/12", "15-01-2024", "2024-01-20", "20240125"],
    "Customer": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Amount": [250, 400, 150, 300, 500]
}

df = pd.DataFrame(data)
df

# Converting Text Dates to Real Dates
df['OrderDate'] = pd.to_datetime(df['OrderDate'], errors='coerce')  # we used errors='coerce' to ignore errors in output
# df
# df.info()  # Check this after conversion of OrderDate that is it really converted to datetime64[ns] or not..
# df['OrderDate'].isna().sum() # Checked how many missing values are there/..

# df.sort_values('OrderDate')
# df['Year'] = df['OrderDate'].dt.year   # we made a new column named 'Year' and put Only the year in it and same for the month column as well...
# df['Month'] = df['OrderDate'].dt.month   # .dt. works only after conversion.
# df
df['OrderDate'] = df['OrderDate'].fillna(pd.Timestamp('2024/01/01'))
df
```

## 2. Setting DateTime Index & Resampling

Converting columns to datetime (pd.to_datetime)

Setting a DateTime index & resampling

```python
import pandas as pd

# 1️⃣ Create a dataset of website visits (string dates + visit counts)
data = {
    "VisitDate": [
        "2024-01-03", "2024/01/08", "15-01-2024",
        "2024-02-02", "2024-02-15",
        "2024/03/01", "2024-03-12", "01-04-2024"
    ],
    "Visitors": [120, 150, 90, 200, 180, 220, 210, 300]
}

df = pd.DataFrame(data)

# 2️⃣ Convert VisitDate to datetime
# Write your code here

# print(df)
df['VisitDate'] = pd.to_datetime(df['VisitDate'], errors='coerce')
# print(df)
# df.info()

# 3️⃣ Set VisitDate as the index
# Write your code here
df = df.set_index('VisitDate')
# print(df)

# 4️⃣ Resample to get total visitors per month
# Write your code here
month = df.resample('ME')['Visitors'].sum()
# print(month)

# 5️⃣ Resample to get average visitors per week
# Write your code here

weekly = df.resample('W')['Visitors'].mean()
print(weekly)
```
