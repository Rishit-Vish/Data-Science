## 1. Applying Custom Functions with `apply()` / `map()`
- `apply()` lets you apply a custom function across rows or columns.
- `map()` is for applying functions element-wise to a Series.
- Both are handy for transformations, calculations, and cleaning data.

```python
import pandas as pd
import numpy as np
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Math": [88, 75, 92, 60, 85],
    "Science": [90, 80, 78, 70, 95]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Adding 5 Marks for every student
df['Math_Bonus'] = df['Math'].apply(lambda x:x+5)
print(df)
```

```python
# Taking Average from both rows
df['Average'] = df.apply(
    lambda row : (row['Math']+ row['Science'])/2,
    axis = 1
)
print(df)
```

```python
# Custome Func
def grade(row):
  avg = (row['Math']+ row['Science']) / 2

  if avg >= 90:
    return 'A'
  elif avg >= 75:
    return 'B'
  else:
    return 'C'
df['Grade'] = df.apply(grade, axis=1)
df
```

```python
df['Final'] = df['Average'].apply(lambda x: "Pass" if x > 50 else "Fail")
df
```

```python
# MAP - works only on 1 column while .apply() works more and is much flexible then map, so just a single example of map
df['Name'] = df['Name'].map(str.upper)  # Map works only on one column, so we can apply the same funcs as of apply, but on 1 column only...
print(df)
```
