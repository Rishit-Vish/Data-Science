## 2. Using `merge()` for SQL-Style Joins
- `merge()` is similar to SQL joins.
- Specify `on` or `left_on`/`right_on` to choose key columns.
- Join types: `inner` (default), `left`, `right`, `outer`.

```python
import pandas as pd

customers = pd.DataFrame({
    "customer_id": [1, 2, 3],
    "name": ["Alice", "Bob", "Charlie"]
})

orders = pd.DataFrame({
    "customer_id": [1, 2, 4],
    "order_amount": [250, 150, 400]
})

result = pd.merge(customers, orders, on="customer_id")
print(result,"\n\n")

left = pd.merge(customers, orders, on='customer_id', how='left')
print(left, "\n\n")

right = pd.merge(customers, orders, on='customer_id', how='right')
print(right,'\n\n')

outer = pd.merge(customers, orders, on='customer_id', how='outer')
print(outer,'\n\n')

inner = pd.merge(customers, orders,on='customer_id', how='inner')
print(inner)
```
