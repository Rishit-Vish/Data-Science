## 1. Concept of Split–Apply–Combine
- **Split** the data into groups based on some criteria.
- **Apply** a function (e.g., sum, mean) to each group.
- **Combine** the results into a new DataFrame or Series.

This is the core idea behind `groupby`.

```python
import pandas as pd

# New dataset: Employees, their Department, and Salary
data = {
    "Employee": ["John", "Sarah", "Mike", "Anna", "Tom", "Laura", "Steve", "Kate"],
    "Department": ["HR", "IT", "IT", "Finance", "Finance", "HR", "IT", "Finance"],
    "Salary": [50000, 70000, 72000, 65000, 60000, 52000, 75000, 58000]
}

df = pd.DataFrame(data)
# group department by salarys
salary = df.groupby('Department')['Salary'].mean()  # This means we used here is called 'Aggregation'.
salary
# department wise employee count
empl = df.groupby("Department")['Employee'].count()
empl
# Aggregation
salary = df.groupby('Department')['Salary'].agg(['mean','max','min','count'])  # This means we used here is called 'Aggregation'.
salary
```

## 2. Using `groupby()` with Aggregation Functions
- Use `groupby('col').mean()` or other functions like sum, count, min, max.
- You can also use `agg()` to pass multiple functions at once.

## 3. Multiple Aggregations on Different Columns
- With `agg()`, you can specify different functions for each column.
- Example: `df.groupby('Dept').agg({'Salary':'mean', 'Age':'max'})`.

```python
data = {
    "Employee": ["John", "Sarah", "Mike", "Anna", "Tom", "Laura", "Steve", "Kate"],
    "Department": ["HR", "IT", "IT", "Finance", "Finance", "HR", "IT", "Finance"],
    "Salary": [50000, 70000, 72000, 65000, 60000, 52000, 75000, 58000],
    "Age": [28, 32, 30, 27, 35, 29, 31, 26]
}

df = pd.DataFrame(data)
df
result = df.groupby('Department').agg(  # for making multiple columns, use dict inside dict
    {
        'Salary' :['mean', 'max', 'min'],
        'Employee' :['count'],
        'Age' :['max','mean']
    }
)
result
```

## 4. Pivot Tables & Crosstab
- `pivot_table()` summarizes data like Excel pivot tables.
- `crosstab()` shows frequency counts of combinations of factors.
- Both are powerful for summarizing and comparing groups.

```python
pivot = df.pivot_table(values='Salary', index='Department', aggfunc="mean")
pivot

pivot = df.pivot_table(values=['Salary','Age'], index='Department', aggfunc=["sum","mean"])  # i don't know whether this two agg func works or not
pivot

# Crosstab -> No of employees by Department and Age group
# Dataframe -> new column AgeGroup column

#pd.cut() -> columns, bins(diff from one column to another one), label (an indicator), this pd.cut() does slicing
df['AgeGroup'] = pd.cut(df['Age'], bins = [25,30,35], labels=["25-30","31-35"])
result = pd.crosstab(df['Department'], df['AgeGroup'])
result
```
