# **Class 1**

##### Matplotlib Basics

```python
x = np.linspace(1,5, 11)
x
y = x ** 2
y
```

```python
plt.plot(x,y)
plt.title('My plot')
plt.xlabel('X axis')
plt.ylabel('Y axis')
```

##### Subplot in Matplotlib

```python
plt.subplot(2,2,1)
plt.plot(x,x)
plt.subplot(2,2,2)
plt.plot(x,x**2)
plt.subplot(2,2,4)
plt.plot(y**2, y)
plt.subplot(2,2,3)
plt.plot(x,y)
```

##### Object Oriented Way

```python
fig = plt.figure(figsize=(4,4), dpi= 100)
axis = fig.add_axes([0.1,0.1,0.7,0.7])
axis.plot(x,y)
axis2 = fig.add_axes([0.3,0.5,0.2,0.2])
axis2.plot(x,y)
```

##### Types of plot

```python
plt.scatter(x,y)
```

```python
plt.hist(x)
```

```python
plt.boxplot(x)
```

##### Saving plot

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.3, 0.3])
ax.plot(x,y)
plt.savefig('Basicplot.png')
```
