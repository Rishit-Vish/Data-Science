Build a dashboard-style summary using multiple visualizations that gives management a quick overview of the business.

The dashboard should help answer questions related to sales performance, profitability, customer behavior, and regional performance.

```python
# Sales over time
plt.figure(figsize=(12,5))

plt.subplot(2,2,1)
monthly_sales = df.groupby('Month Year')['Sales'].sum().sort_index()
plt.plot(monthly_sales.index.astype(str), monthly_sales.values)
plt.xlabel('Month')
plt.ylabel('total sales')
plt.xticks(rotation=45)
plt.xticks(monthly_sales.index[::5].astype(str))


# Profit over time
plt.subplot(2,2,2)
profit_monthly = df.groupby('Month Year')['Profit'].sum().sort_index()
plt.plot(profit_monthly.index.astype(str), profit_monthly.values)
plt.xlabel('Month')
plt.ylabel('Profit')
plt.xticks(profit_monthly.index[::5].astype(str))
plt.xticks(rotation = 45)

# Regional Performance By Sales
plt.subplot(2,2,3)
region_sales = df.groupby('Region')['Sales'].sum().sort_index()
plt.bar(region_sales.index, region_sales.values)
plt.xlabel('Region')
plt.ylabel('Sales')

# Category Performance over time
plt.subplot(2,2,4)
cat_sales = df.groupby('Category')['Profit'].sum().sort_values()
plt.bar(cat_sales.index, cat_sales.values)
plt.tight_layout()
```

```python
# Overall Business health
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
sales_time = df.groupby('Month Year')['Sales'].sum().sort_index()
plt.plot(sales_time.index.astype(str), sales_time.values)
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(sales_time.index[::5].astype(str), rotation=45)
plt.title('Sales by time')
plt.grid(alpha=0.3)

plt.subplot(1,2,2)
profit_time = df.groupby('Month Year')['Profit'].sum().sort_index()
plt.plot(profit_time.index.astype(str), profit_time.values)
plt.xlabel('Month')
plt.ylabel('Total Profit')
plt.xticks(profit_time.index[::5].astype(str), rotation=50)
plt.title('Profit by time ')
plt.grid(alpha=0.3)


plt.tight_layout()
plt.show()
```

```python
# Where money comes from?
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
region_profit = df.groupby('Region')['Profit'].sum().sort_values()
plt.bar(region_profit.index.astype(str), region_profit.values)
plt.xlabel('Region')
plt.ylabel('Profit')
plt.title('Profit by region')

plt.subplot(1,2,2)
category_profit = df.groupby('Category')['Profit'].sum().sort_values()
plt.bar(category_profit.index.astype(str), category_profit.values)
plt.xlabel('Category')
plt.ylabel('Profit')
plt.title('Profit by category')

plt.tight_layout()
```

```python
# Growth opportunities
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
disc_prof = df.groupby('Discount')['Profit'].sum().sort_values()
plt.plot(disc_prof.index.astype(str),disc_prof.values)
plt.grid(alpha=0.3)
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.title('Profit By Discount')

plt.subplot(1,2,2)
low_prof_region = df.groupby('Region')['Profit'].sum().sort_values()
plt.bar(low_prof_region.index.astype(str), low_prof_region.values)
plt.xlabel('Region')
plt.ylabel('Profit')
plt.title('Profit by Region')

plt.tight_layout()
```

```python
# Customers

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
customer_profit = df.groupby('Customer Name')['Profit'].sum().sort_values(ascending=False).head(10)
plt.barh(customer_profit.index.astype(str), customer_profit.values)
plt.xlabel('Customer')
plt.ylabel('Profit')
plt.title('Top 10 Customers by Profit')


plt.subplot(2,2,2)
strong_segment = df.groupby('Segment')['Profit'].sum().sort_values(ascending=False)
plt.bar(strong_segment.index.astype(str), strong_segment.values)
plt.xlabel("Segment")
plt.ylabel('Profit')
plt.title('Profit by Segment')

plt.tight_layout()
```
